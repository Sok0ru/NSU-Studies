## Задание 13

Найдите евклидовы расстояния между каждой парой точек из `coords`. Результат `distances` представьте матрицей 8×8: элемент `[i, j]` — расстояние между точками `i` и `j`.

```python
distances = np.zeros((len(coords), len(coords)))
for row in range(len(coords)):
    for col in range(len(coords)):
        dx = coords[row][0] - coords[col][0]
        dy = coords[row][1] - coords[col][1]
        distances[row][col] = np.sqrt(dx ** 2 + dy ** 2)
print(distances)
```

## Задание 14

Напишите функцию `add_ones(vector, index)`, которая увеличивает соответствующие элементы `vector` на 1 для каждой позиции из `index`. Если индекс повторяется, прибавление выполняется несколько раз. Функция должна возвращать изменённый массив; исходный массив можно изменять или копировать.

```python
def add_ones(vector, index):
    result = np.array(vector, copy=True)
    for position in index:
        result[position] += 1
    return result


print(add_ones(np.array([0, 0, 0, 0]), [1, 1, 3]))
```

## Задание 15

Напишите функцию `insert_zeros(vector, count)`, возвращающую новый вектор с `count` последовательными нулями **между** соседними значениями исходного `vector` (без нулей до первого и после последнего элемента).

```python
def insert_zeros(vector, count):
    result = []
    for position in range(len(vector)):
        result.append(vector[position])
        if position < len(vector) - 1:
            for _ in range(count):
                result.append(0)
    return np.array(result, dtype=np.asarray(vector).dtype)


print(insert_zeros(np.array([1, 2, 3]), 2))
```

## Задание 16

Напишите функцию `moving_average(vector, d_size_window)`, вычисляющую скользящее среднее с окном размера `d_size_window`. Результат должен иметь длину `len(vector) - d_size_window + 1`. Предполагается, что `1 <= d_size_window <= len(vector)`.

```python
def moving_average(vector, d_size_window):
    result = []
    for start in range(len(vector) - d_size_window + 1):
        window = vector[start:start + d_size_window]
        result.append(np.mean(window))
    return np.array(result)


print(moving_average(np.array([1, 2, 3, 4, 5]), 3))
```

## Задание 17

Напишите функцию `most_common(arr)`, которая находит наиболее часто встречающееся значение в непустом массиве `arr`. Если таких значений несколько, возвращайте наименьшее.

```python
def most_common(arr):
    values, counts = np.unique(arr, return_counts=True)
    max_index = np.argmax(counts)
    return values[max_index]


print(most_common(np.array([3, 1, 3, 1, 2])))
```

## Задание 18

Даны два массива одинаковой длины: реальные и синтетические данные. Напишите функцию `random_select(real_data, synthetic_data, p)`, которая **независимо для каждого элемента** с вероятностью `p` берёт синтетическое значение, а с вероятностью `1-p` — реальное. Порядок позиций сохраняется; функция возвращает новый массив.

```python
def random_select(real_data, synthetic_data, p):
    result = []
    for position in range(len(real_data)):
        if np.random.random() < p:
            result.append(synthetic_data[position])
        else:
            result.append(real_data[position])
    return np.array(result)


print(random_select(np.array([1, 2, 3, 4, 5]),
                    np.array([10, 20, 30, 40, 50]), 0.5))
```
