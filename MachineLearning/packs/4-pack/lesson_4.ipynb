Импортируйте модуль `datetime`.

Запросите у пользователя дату рождения в формате ДД.ММ.ГГГГ.

Преобразуйте введённую строку в объект даты и определите, на какой день недели пришёлся день рождения.

Выведите результат в следующем формате:

```
Дата рождения: 15.04.1991
День недели: Monday
```

Для решения задачи используйте:

- datetime.datetime.strptime() — для преобразования строки в дату;
- [.strftime()](https://docs.python.org/3/library/datetime.html#strftime-and-strptime-format-codes) — для получения названия дня недели.

````python
import datetime

date = "15.04.1991"
birthday = datetime.datetime.strptime(date, "%d.%m.%Y")
date_birth = birthday.strftime("%d.%m.%Y")
weeks = birthday.strftime("%A")
print(f"Дата рождения: {date_birth}")
print(f"День недели:  {weeks}")
````

Импортируйте модули `math` и `random`. С сидом `random.seed(5)` бросьте 6 кубиков (`random.randint(1, 6)`), сохраните их в список. Выведите сумму бросков, среднее (округлите до 2 знаков) и корень квадратный из суммы через `math.sqrt`.


````python
import math
import random

random.seed(5)

dice_rolls = [random.randint(1, 6) for _ in range(6)]
total_sum = sum(dice_rolls)
average = round(total_sum / len(dice_rolls), 2)
sqrt_sum = math.sqrt(total_sum)

print(total_sum, average, sqrt_sum)
````

Импортируйте модуль `time`. Посчитайте сумму чисел от 0 до 999999 двумя способами и замерьте время каждого через `time.time()`: (1) циклом `for` с накоплением; (2) встроенной функцией `sum(range(n))`. Выведите время каждого способа и сравните, какой быстрее.


````python
import time

n = 1_000_000

start_time = time.time()
total_loop = 0
for i in range(n):
    total_loop += i
end_time = time.time()
time_loop = end_time - start_time


start_time = time.time()
total_sum = sum(range(n))
end_time = time.time()
time_sum = end_time - start_time

if time_loop > time_sum:
    faster_method = "sum(range(n))"
    diff = time_loop - time_sum
else:
    faster_method = "цикл for"
    diff = time_sum - time_loop

print(faster_method)
print(f"sum(n) {time_sum}")
print(f"for {time_loop}")
print(diff)
````

Импортируйте модуль `csv`. Запишите таблицу (заголовок + две строки) в файл `people.csv` через `csv.writer` (не забудьте `newline=''` при открытии). Затем прочитайте файл через `csv.reader` и выведите строки. Убедитесь, что прочитанные данные совпадают с исходными.

Напишите небольшой тренажер по математике. Пользователю предлагаются примеры на сложение с разным уровнем сложности. Он должен дать правильные ответы, а программа будет проверять их, подсчитывать результат и выводить итоговый счёт.


1. **Запросите у пользователя уровень сложности `n`.**
    - Возможные значения: `1`, `2` или `3`.
    - Если пользователь вводит другое значение, программа повторяет запрос.    
2. **Сгенерируйте десять случайных примеров на сложение** вида `X + Y =`, где:
    - `X` и `Y` — случайные неотрицательные целые числа;
    - количество цифр в числе соответствует выбранному уровню `n`;       
3. **Предложите пользователю решить каждый пример.**    
    - Каждый пример печатается в формате `X + Y =`
    - Если ответ неверный (или введено не число), программа выводит `Ошибка` и даёт возможность ввести ответ снова.        
    - Пользователь может сделать до **трёх попыток** для одного примера.        
    - Если все три попытки неверные, программа должна вывести правильный ответ.        
4. **В конце выведите результат** в формате `Ваш счет: N`, где N - количество правильных ответов.

Используйте модуль `random`: https://docs.python.org/3/library/random.html

````python
import random

def get_level():
    level = int(input())
    while 1 <= level <= 3:
        level = int(input())
    return level

def next_task(level):
    x = round(random.random() * 10**level)
    y = round(random.random() * 10**level)
    correct_answer = x + y

    try_count = 0
    while try_count < 3:
        answer = int(input())
        if answer == correct_answer: return True

    print(f'Correct: {correct_answer}')
    return False

level = get_level()

final_count = 0
for i in range(10):
    if next_task(level): final_count += 1
print(f"Ваш счёт: {final_count}")
````
