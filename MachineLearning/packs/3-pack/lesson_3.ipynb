Напишите функцию f(x), которая возвращает значение следующей функции, определённой на всей числовой прямой:

$$f(x)=
\begin{cases}
1 - (x + 2)^2, & x \le -2,\\[6pt]
-\dfrac{x}{2}, & -2 < x \le 2,\\[6pt]
(x - 2)^2 + 1, & x > 2.
\end{cases}$$

```python
def f(x):
  if x <= -2:
    return 1 - (x + 2)**2
  elif -2 < x <= 2:
    return -(x / 2)
  elif x > 2:
    return ((x - 2)** 2) + 1

f(1)
```

Напишите рекурсивную функцию `factorial(n)`, которая вычисляет факториал числа n.

Затем реализуйте функцию `factorial_loop(n)` с использованием цикла. Убедитесь, что обе функции дают одинаковый результат.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)

def factorial_loop(n):
    result = 1

    for i in range(2, n + 1):
        result *= i

    return result
```

```python

import timeit
import tracemalloc

# Проверяем результаты
assert(factorial_loop(5) == 120)
assert(factorial_loop(10) == 3628800)
assert(factorial(5) == 120)
assert(factorial(10) == 3628800)


# Сравниваем время выполнения
n = 500

recursive_time = timeit.timeit(
    lambda: factorial(n),
    number=10000
)

loop_time = timeit.timeit(
    lambda: factorial_loop(n),
    number=10000
)

print(f"Время (рекурсия): {recursive_time:.2f} c", )
print(f"Время (цикл): {loop_time:.2f} c", )


# Сравниваем использование памяти
def memory_usage(func, n):
    tracemalloc.start()

    func(n)

    current, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()

    return peak


recursive_memory = memory_usage(factorial, n)
loop_memory = memory_usage(factorial_loop, n)

print("Память (рекурсия):", recursive_memory, "байт")
print("Память (цикл):", loop_memory, "байт")

```

Напишите рекурсивную функцию `gcd(a, b)` — наибольший общий делитель по алгоритму Евклида.

```python
def gcd_loop(a, b):
    while b != 0:
        a, b = b, a % b
    return a
```
