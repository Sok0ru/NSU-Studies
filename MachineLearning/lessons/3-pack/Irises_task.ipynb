- Ширина наружной доли околоцветника (англ. sepal width);
- Длина внутренней доли околоцветника (англ. petal length);

## Часть 1

В файле `train.csv` содержится информация о ширине околцветников (в этом задании мы не используем длину) некоторых ирисов, описанных Фишером.

Каждая строка в файле имеет вид:

sepal_width,petal_width,iris_type

где sepal_width и petal_width представлены числами с плавающей запятой, iris_type - строка (вид ириса).

Сохраните информацию о всех ирисах из файла в список `known_irises`, информация о каждом ирисе должна хранится в словаре (dict) со следующими ключами:
- type
- sepal_width
- petal_width

```python
import csv
from math import sqrt, inf

known_irises = []
with open('train.csv', 'r') as f:
  for line in f:
    data = line.strip().split(',')
    iris = {'type': data[2],
            'sepal_width': float(data[0]),
            'petal_width': float(data[1])
    }
    known_irises.append(iris)
print(known_irises)
```

```python
assert len(known_irises) == 112, f"Длина списка known_irises должна быть 112, а не {len(known_irises)}"
assert set(known_irises[0].keys()) == {'type', 'sepal_width', 'petal_width'}, f"Ключи словаря known_irises должны быть 'type', 'sepal_width', 'petal_width', а не {set(known_irises[0].keys())}"
```

## Часть 5

Как определить, какой вид у неизвестных ирисов? Возьмем любую серую точку (ирис с неизвестным видом) и посмотрим, какой вид (цвет) у ближайшей точки (или нескольких ближайших). Скорее всего у нашего неизвестного ириса будет такой же вид.

Примечание: Этот метод называется методом k ближайших соседей (где k - количество рассматриваемых соседей)

### Часть 5.1

Чтобы найти ближайший ирис, нужно уметь находить расстояние между двумя точками на плоскости.

Евклидово расстояние между точками $(x1,y1)$ и $(x2,y2)$ вычисляется по следующей формуле:

$$dist(x1,y1,x2,y2)=\sqrt{(x1-x2)^2+(y1-y2)^2}$$

Напишите функцию, вычисляющую это расстояние.

```python
import math
def find_distance(x1, y1, x2, y2):
    return math.sqrt((x1 - x2)**2 + (y1 - y2)**2)
```

```python
assert find_distance(0, 0, 3, 4) == 5.0, f"Функция find_distance работает неправильно"
```

### Часть 5.2

Теперь найдем ближайшего соседа (ирис с известным видом, до которого расстояние минимально).

Напишите функцию `find_nearest`, которая для ириса с координатами (sepal_width, petal_width) ищет ближайший среди `irises`. Аргумент функции `irises` должен иметь такой же формат как и `known_irises`, т.е. это список словарей, каждый из которых содержит три ключа: type, sepal_width и petal_width.

```python
def find_nearest(sepal_width, petal_width, irises):
    nearest_iris = None
    nearest_distance = math.inf
    for iris in irises:
        distance = find_distance(sepal_width, petal_width, iris['sepal_width'], iris['petal_width'])
        if distance < nearest_distance:
            nearest_iris = iris
            nearest_distance = distance

    return nearest_iris
```

```python
assert find_nearest(3.0, 1.0, known_irises)['type'] == 'Iris-versicolor', f"Функция find_nearest работает неправильно"
```

## Часть 6

Предскажем вид для каждого неизвестного ириса. Следующий код использует функцию find_nearest, чтобы найти ближайшего соседа, и делает предсказание, используя вид этого ближайшего ириса.

```python
predictions = []
for iris in unknown_irises:
    nearest = find_nearest(iris['sepal_width'], iris['petal_width'], known_irises)
    predictions.append(nearest['type']) # предсказанный вид ириса
```

Сохраните предсказания в файл `predictions.csv`: одна строка - одно предсказание (в том же порядке, что в списке predictions).

```python
import csv
with open('predictions.csv', 'w', newline='') as f:
  csv_file = csv.writer(f)
  csv_file.writerows([[predication] for predication in predictions])
```

## Часть 7

Оценим качество наших предсказаний.
В файле `answers.csv` содержаться правильные виды ирисов из файла `test.csv` (в том же порядке).

Посчитайте процент правильных предсказаний среди `predictions` и сохраните в переменную accuracy

```python
answers_csv = []
with open('answers.csv', 'r') as f: answers_csv = list(csv.reader(f))

predictions_csv = []
with open('predictions.csv', 'r') as f: predictions_csv = list(csv.reader(f))

equal_count = 0
for i, predication in enumerate(predictions_csv):
    if predication == answers_csv[i]: equal_count += 1

accuracy = equal_count / len(answers_csv)
```
